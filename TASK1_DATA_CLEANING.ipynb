```python
import pandas as pd
import numpy as np
import seaborn as sns
```

```python
df = pd.read_csv("titanic.csv")
```

```python
df
```

```python
df.isnull().sum()
```

```python
df.head(5)
```

```python
df.columns
```

```python
df.drop(columns=["PassengerId","SibSp","Parch","Pclass","Embarked"],inplace=True)
```

```python
df
```

```python
df.head(4)
```

```python
df.isnull().sum()
```

```python
df["Age"] = df["Age"].fillna(df["Age"].median())
```

```python
df["Cabin"] = df["Cabin"].fillna("Unknown")
```

```python
df.isnull().sum()
```

```python
df
```

```python
df = pd.get_dummies(df,columns=["Sex"],drop_first=True,dtype="int64")
```

```python
df.head()
```

```python
print(df.head())
print(df.shape)
print(df.dtypes)
```

```python
print("Duplicates before cleaning:", df.duplicated().sum())

df.drop_duplicates(inplace=True)

print("Duplicates after cleaning:", df.duplicated().sum())
```

```python
df
```

```python
df.to_csv("titanic_cleaned.csv", index=False)

from google.colab import files
files.download("titanic_cleaned.csv")
```

```python
report = pd.DataFrame({
    "Data Type": df.dtypes.astype(str),
    "Missing Values": df.isnull().sum(),
    "Unique Values": df.nunique()
})

display(report)

print("Final rows and columns:", df.shape)
print("Remaining duplicate rows:", df.duplicated().sum())
print("Total missing cells:", df.isnull().sum().sum())

report.to_csv("data_quality_report_after.csv")
files.download("data_quality_report_after.csv")
```
